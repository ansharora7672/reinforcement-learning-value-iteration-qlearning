```python
import numpy as np
```

```python
# =====================================================================
# We have 4 states (s0, s1, s2, s3) and 4 possible actions (a0, a1, a2, a3).
# We will use numpy arrays to easily map Transitions (P) and Rewards (R).
# Array format: [current_state, action, next_state]
# =====================================================================
```

```python
num_states = 4
num_actions = 4
```

```python
# Initializing empty grids for Probabilities (P) and Rewards (R)
P = np.zeros((num_states, num_actions, num_states))
R = np.zeros((num_states, num_actions, num_states))
```

```python
# --- STATE s0 ---
# Action a0: 40% chance to stay in s0 (0 pts), 60% chance to move to s1 (+10 pts)
P[0, 0, 0] = 0.4; R[0, 0, 0] = 0
P[0, 0, 1] = 0.6; R[0, 0, 1] = 10
# Action a1: 100% chance to move to s1 (0 pts)
P[0, 1, 1] = 1.0; R[0, 1, 1] = 0
```

```python
# --- STATE s1 ---
# Action a1: 100% chance to stay in s1 (0 pts)
P[1, 1, 1] = 1.0; R[1, 1, 1] = 0
# Action a2: 100% chance to move to s0 (-10 pts)
P[1, 2, 0] = 1.0; R[1, 2, 0] = -10
# Action a3: 80% chance to move to s2 (0 pts), 20% chance to move to s3 (-10 pts)
P[1, 3, 2] = 0.8; R[1, 3, 2] = 0
P[1, 3, 3] = 0.2; R[1, 3, 3] = -10
```

```python
# --- STATE s2 ---
# Action a0: 30% chance to move to s0 (-10 pts), 70% chance to move to s3 (-10 pts)
P[2, 0, 0] = 0.3; R[2, 0, 0] = -10
P[2, 0, 3] = 0.7; R[2, 0, 3] = -10
# Action a2: 80% chance to move to s1 (0 pts), 20% chance to stay in s2 (+10 pts)
P[2, 2, 1] = 0.8; R[2, 2, 1] = 0
P[2, 2, 2] = 0.2; R[2, 2, 2] = 10
```

```python
# --- STATE s3 ---
# Action a1: 10% chance to move to s1 (-10 pts), 90% chance to stay in s3 (+10 pts)
P[3, 1, 1] = 0.1; R[3, 1, 1] = -10
P[3, 1, 3] = 0.9; R[3, 1, 3] = 10
```

```python
# Dictionary defining which actions are actually available in each state
# based on the blue diamonds in the diagram.
possible_actions = {
    0: [0, 1],       # In s0, can take a0, a1
    1: [1, 2, 3],    # In s1, can take a1, a2, a3
    2: [0, 2],       # In s2, can take a0, a2
    3: [1]           # In s3, can take a1
}
```

```python
# =====================================================================
# INITIALIZING Q-VALUES AND HYPERPARAMETERS
# =====================================================================

# Initializing a 4x4 Q-value table with negative infinity
Q_values = np.full((num_states, num_actions), -np.inf)

# Setting the valid State-Action pairs to 0.0 to act as a blank slate
for state, actions in possible_actions.items():
    Q_values[state, actions] = 0.0

gamma = 0.90    #learning rate gamma factor
iterations = 50 # Number of sweeps for Value Iteration
```

```python
# =====================================================================
# VALUE ITERATION (APPLYING THE BELLMAN OPTIMALITY EQUATION)
# =====================================================================

for i in range(iterations):
    Q_prev = Q_values.copy() # Copy old values
    
    for s in range(num_states):
        for a in possible_actions[s]:
            q_value_sum = 0
            
            # Loop through all possible next states (s')
            for sp in range(num_states):
                # Grab Probability and Reward from our grids
                prob = P[s, a, sp]
                reward = R[s, a, sp]
                
                # If there's a chance to go to this state, calculate its value
                if prob > 0:
                    # Future expected reward: Gamma * Max(Q_next)
                    future_reward = gamma * np.max(Q_prev[sp])
                    # Apply Bellman formula chunk
                    q_value_sum += prob * (reward + future_reward)
            
            # Update the Q-table
            Q_values[s, a] = q_value_sum
```

```python
# =====================================================================
# EXTRACTING THE OPTIMAL POLICY
# =====================================================================

optimal_policy = np.argmax(Q_values, axis=1)

print("Final calculated Q-Values table:")
print(np.round(Q_values, 2))
print("\nOptimal action to take for each state:")
for state in range(num_states):
    print(f"State s{state} -> Action a{optimal_policy[state]}")
```
